## Adjusted Close vs. Unadjusted Close

<mark>Q: What is the difference between the adjusted close and the unadjusted close?</mark>

**A:** The unadjusted close is the actual traded price at the close. The adjusted close is back-adjusted for dividends and splits, so its percentage changes reflect the full return to a holder. Before each ex-dividend date, historical adjusted prices are scaled down, so Adj Close sits below Close earlier in the sample.

# Price Return

* The daily price return is the percentage change in the price of a security today relative to its price yesterday:

$$\frac{P_t - P_{t-1}}{P_{t-1}}$$

* Calculate the price return by defintion:

<mark>Q: How to calculate the daily price return by definition?</mark>

**A:** Divide today's close minus yesterday's close by yesterday's close: `(Close - Close.shift(1)) / Close.shift(1)`, which is the same as `Close.pct_change()`.

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Extract the `"Close"` column of `df` into a variable called `Close`.

```python
Close = df["Close"]
```

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Calculate the daily price return by definition, `(Close - Close.shift(1)) / Close.shift(1)`, and store it in `price_return`.

```python
price_return = (Close - Close.shift(1)) / Close.shift(1)
```

# Total Return

# Log Return

<mark>Q: What are log returns?</mark>

**A:** Log returns are $\ln(P_t/P_{t-1})$, the continuously compounded return. They are additive over time (multi-period log return = sum of daily log returns) and approximately equal to simple returns when returns are small.

* The returns calculated in the preceding section are called arithmetic returns or simple returns.
* The logarithmic return, or log return, is calculated as:

$$r_t = \ln\frac{P_t}{P_{t-1}} = \ln P_t - \ln P_{t-1}$$

## Calculate Log Return

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Calculate the log return as `np.log(Adjusted).diff()` and store it in `log_return`.

```python
log_return = np.log(Adjusted).diff()
```

# Cumulative Return

<mark>Q: How to calculate the cumulative return from the daily returns?</mark>

**A:** Add 1 to each daily return to get gross returns, then take the cumulative product: `(1 + R).cumprod()`.

* Taking the product of the daily gross returns yields the cumulative return:

$$R_{1\text{ to }T} = (1+R_1)(1+R_2)\cdots(1+R_T)$$

## Calculating Gross Return

### QQ Plot

* Given two samples, a QQ plot visually checks whether they share the same distribution.
* It plots the quantiles of one sample against the quantiles of the other.
* If both come from the same distribution, the points form a roughly straight line.

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Make a QQ plot of the returns against a normal sample with the same mean and standard deviation. Uncomment and run the following code.

### Skewness

* Skewness measures the asymmetry of a distribution. A left-skewed distribution has a long left tail; a right-skewed distribution has a long right tail.
* Skewness is defined as:

$$\text{skewness} = \text{Average}\left[\frac{(R - \bar{R})^{3}}{\hat{\sigma}^{3}}\right]$$

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Calculate the skewness of `r` with `.skew()`.

```python
r.skew()
```

### Kurtosis

<mark>Q: What does positive (excess) kurtosis tell us about the risk of extreme returns?</mark>

**A:** Positive excess kurtosis means fatter tails than a normal distribution. Extreme gains and losses happen more often than a normal model predicts, so normal-based risk measures understate tail risk.

* Kurtosis measures how the tails behave relative to a normal distribution:

$$\text{kurtosis} = \text{Average}\left[\frac{(R - \bar{R})^{4}}{\hat{\sigma}^{4}}\right] - 3$$

* Positive (excess) kurtosis indicates a fat-tailed, *leptokurtic* distribution. In pandas, `.kurt()` returns the excess kurtosis.

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Calculate the (excess) kurtosis of `r` with `.kurt()`.

```python
r.kurt()
```

## Return Predictability

### Can Past Returns Predict Future Returns?

<mark>Q: How to test whether past returns predict future returns?</mark>

**A:** Scatter $r_t$ against $r_{t-1}$ and compute the autocorrelation (and the ACF across lags). Autocorrelations near zero, inside the confidence bands, indicate little linear predictability.

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Let's plot $r_t$ against its lag $r_{t-1}$. Uncomment and run the following code.

```python
plt.scatter(r.shift(1), r)
plt.xlabel("$r_{t-1}$")
plt.ylabel("$r_t$")
plt.title("Return vs. Its Lag")
plt.show()
```

### Autocorrelation

* One way to test predictability is to compute the **autocorrelation**:

**A:** The VIX is the CBOE index of the market's expected 30-day volatility of the S&P 500, derived from S&P 500 option prices and quoted in annualized percentage points.
