# Downsampling

To downsample the daily data to monthly frequency, we can use the `resample()` method and specify how to aggregate the OHLC and Volume data for each month.

<mark>Q: How to convert daily prices to monthly prices?</mark>

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Downsample the data to monthly frequency. Uncomment the code and run.

```python
df_monthly = df.resample('ME').agg({
    'Open': 'first',
    'High': 'max',
    'Low': 'min',
    'Close': 'last',
    'Volume': 'sum'
})
```

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Check the head of `df_monthly`.

```python
df_monthly.head()
```

## Merging the two time series

We can merge the 'Close' prices of `df` and `GSPC` into a single DataFrame using the `join()` method, as they share a common index (the Date).

<mark>Q: How to combine two price series that share a date index?</mark>

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Uncomment and run the given code below. It uses `.join()` to join the `Close` column of `df` with the `Close` column of `GSPC` on their shared date index, with `lsuffix='_i'` and `rsuffix='_m'`, and stores the result in `merged_df`.

```python
# Join the two DataFrames based on their index
merged_df = df[['Close']].join(GSPC[['Close']],
                               lsuffix='_i',
                               rsuffix='_m')
```

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Check the head of `merged_df`.

```python
merged_df.head()
```

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Use `.rename()` with `inplace=True` to rename `Close_i` to `p_i` and `Close_m` to `p_m` in `merged_df`, then display its first five rows.

```python
merged_df.rename(columns={'Close_i': 'p_i', 'Close_m': 'p_m'}, inplace=True)
merged_df.head()
```

<mark>Q: How to plot two price series on one chart?</mark>

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Plot both columns of `merged_df` on one chart, with the title `"Close Price"`.

```python
merged_df.plot(title="Close Price")
```

## Normalization

To normalize the prices to start at 1, divide each series by its first value.

<mark>Q: How to rescale two price series so that both start at 1?</mark>

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Create a column `p_i_1` in `merged_df`: the column `p_i` divided by its first value, `merged_df['p_i'].iloc[0]`.

```python
merged_df['p_i_1'] = merged_df['p_i'] / merged_df['p_i'].iloc[0]
```

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ In the same way, create a column `p_m_1`: the column `p_m` divided by its first value.

```python
merged_df['p_m_1'] = merged_df['p_m'] / merged_df['p_m'].iloc[0]
```

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Inspect the first five rows of `merged_df`.

```python
merged_df.head()
```

$\bbox[lightblue, 3pt]{\color{black}{\text{To do:}}}$ Plot the columns `p_i_1` and `p_m_1` of `merged_df` together, with the title `"Normalized Close Price"`.

```python
merged_df[['p_i_1', 'p_m_1']].plot(title="Normalized Close Price")
```
